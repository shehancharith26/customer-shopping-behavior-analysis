```python
import pandas as pd
import numpy as np

df = pd.read_csv('shopping_trends.csv')

df.head()
```

```python
# 1. Column Names
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('(', '', regex=False)
    .str.replace(')', '', regex=False)
)

# 2. Duplicate rows
df.drop_duplicates(inplace=True)

# 3. Feature Engineering: Age Groups
age_bins = [17, 25, 40, 60, 100]
age_labels = [
    'Gen Z (18-25)',
    'Millennials (26-40)',
    'Gen X (41-60)',
    'Seniors (61+)',
]
df['age_group'] = pd.cut(df['age'], bins=age_bins, labels=age_labels)

# 4. Feature Engineering: Spending Tiers

target_col = (
    'purchase_amount_usd'
    if 'purchase_amount_usd' in df.columns
    else [c for c in df.columns if 'purchase' in c][0]
)

q1 = df[target_col].quantile(0.33)
q2 = df[target_col].quantile(0.66)


def classify_spender(amount):
  if amount <= q1:
    return 'Low Spender'
  elif amount <= q2:
    return 'Medium Spender'
  else:
    return 'High Spender'


df['spending_tier'] = df[target_col].apply(classify_spender)

df.to_csv('cleaned_shopping_trends.csv', index=False)
print(
    "✅ Cleaning Completed Successfully! 'cleaned_shopping_trends.csv' is"
    ' ready.'
)
```
